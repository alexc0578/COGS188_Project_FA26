# Abstract 

This project will implement and benchmark three reinforcement learning methods to solve a robotic manipulation task using an SO-101 robotic arm in MuJoCo simulation. Our goal is to compare task success, sample efficiency, and computational performance. We will first develop and train in simulation, then attempt to transfer our policies to a physical robot. Success for this task will include a working MuJoCo manipulation environment; PPO, PPO + SGS, and TD-MPC2 implementations; Quantitative results and learning curves; and if time/resources permit, a real-world SO-101 demonstration.

# Background

Manipulation is one of the hardest unsolved problems in robotics, with traditional approaches requiring engineers to design controllers, specify object models, and plan trajectories for each task--which works in controlled environments but breaks down when objects, lighting, or positions change. Reinforcement learning offers an alternative to this workflow: instead of being pre-programmed with instructions for how to perfrom a task, the robot can learn its own strategy by trial and error while guided by a reward signal. This could allow a single learning method to produce robotic skills for many situations without explicit engineering.

This research matters because many valuable applications such as warehouse logistics, manufacturing, and household assistance depend on the robots being able to handle varied, unstructured objects reliably.

Prior research in robotic manipulation rests on key developments such as the development of MuJoCo--a physicds engine that was designed for fast, accurate simulation of contact<a name="MuJoCo"></a>[<sup>[5]</sup>](#MuJoConote). Because the software is fast and stable, researchers can run millions of simulations that would be impossible to try on a real robot.

Another issue that has been addressed was the gap between simulations and reality. A policy that was trained in simulation often fails on a real-world robot due to friction or motor delays that the simulator doesn't perfectly capture. Domain randomization<a name="Tobin"></a>[<sup>[6]</sup>](#Tobinnote) addressed this by adding deliberate variation to properties such as those during training, so the policy would learn to handle a wide range of conditions and helps generalize to the real world.

This project will use the same open source, low-cost hardware as prior work (SO-101, MuJoCo) and will provide a consistent evaluation framework and concrete metrics of robotic arm's performance on the manipulation task. And if time permits, this project will provide a numeric reference point for the difference between simulated and real-world success rate when tested on hardware.

# Data

**Simulation**
- The SO-101<a name="HuggingFace"></a>[<sup>[2]</sup>](#HuggingFacenote) is the robotic arm that will be used in our project's manipulation task.
- MuJoCo is an open-source, advanced physics engine that the simulations will be run on.

**Algorithms**
- **Proximal Policy Optimization (PPO)** is an architecture that improves agent's training stability by avoiding policy updates that are too large<a name="Simonini"></a>[<sup>[4]</sup>](#Simonininote), and is a technique widely used in RL.
- **PPO + SGS** (Success Guided Sampling) is a new method that has shown impressive performance. SGS is standard PPO with a small outer loop which changes only which task configuration an environment resets too. It trains on the task configurations that are neither too easy nor hard for the current policy.<a name="Zhang"></a>[<sup>[3]</sup>](#Zhangnote)
- TD-MPC2 is a model-based reinforcement learning algorithm that performs local trajectory optimization in the latent space of a learned implicit (decoder-free) world model<a name="Hansen"></a>[<sup>[1]</sup>](#Hansennote). It was developed by PhDs (Nicklas Hansen, Hao Su, and Xiaolong Wang) at UCSD.

# Evaluation Metrics

**Task Performance**: Success rate of task-specific performance score

$\frac{1}{N}\sum_i s(\tau_i)$
- With $s(\tau_i)$ being the binary success indicator in episode i (1 if success condition is met, 0 otherwise)
- The sum of # successful episodes normalized by number of total episodes.

**Sample Efficiency**: Performance relative to environment interactions

$\;N_x = \min\{n : J(n)\ge x\}$
- J(n) being performance after n steps
- x being target performance threshold chosen (e.g. 0.9)
- $\min\{n : J(n)\ge x\}$ being the smallest number of interactions which performance reaches the x value.

**Compute Efficiency**: Training time and computational resources

$\;\eta_{\text{compute}} = J/C$
- C being total compute cost measured in GPU hours
- J being the final performance achieved at the end of training

**Generalization**: Performance on unseen initial configurations
$\;\Delta_{\text{gen}} = J_{\text{train}} - J_{\text{test}}$
- $J_{\text{train}}$ being performance measured on configurations seen during training.
- $J_{\text{test}}$ being performance on configurations in t=the test set (unseen object poses, positions, etc.)

# Ethics & Privacy

**Broader Usage** Though our intention is to evaluate the performance of a low-cost robotic arm in simulation, we realize this research could have harmful applications. We want our robotic arm research to be used for good--making lives easier in warehouses or administering aid in situations where humans can't help--and will not give our results to Lockheed Martin.
